## A tiny “analysis toolbox” we’ll reuse

These helper functions:
- set random seeds (reproducibility)
- make quick plots
- add noise in a controlled way


```python
import numpy as np
import matplotlib.pyplot as plt

def set_seed(seed: int = 42):
    np.random.seed(seed)

def imshow(img, title=None, vmin=None, vmax=None):
    plt.figure(figsize=(6,5))
    plt.imshow(img, origin="lower", vmin=vmin, vmax=vmax)
    plt.colorbar(label="counts")
    if title: plt.title(title)
    plt.xlabel("x [pix]")
    plt.ylabel("y [pix]")
    plt.tight_layout()
    plt.show()

def add_gaussian_noise(x, sigma):
    return x + np.random.normal(0.0, sigma, size=np.shape(x))

set_seed(7)
```
